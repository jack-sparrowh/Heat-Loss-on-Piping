This example showcases, how user can provide imput in SI units.

```python
from py_files.data_prep import *
from py_files.data_utils import *
from py_files.mainframe import *
from py_files.data_example import *
import matplotlib.pyplot as plt
```

```python
user_array = user_table_example()[3:]
properties_table_raw = properties_table_example()
```

```python
for i, unit in enumerate(['imperial', 'si']):
    print(unit)
    print('-'*10)
    print(user_array[i], '\n')
```

```python
for i, unit in enumerate(['imperial', 'si']):
    print(unit)
    print('-'*10)
    print(properties_table_raw[i], '\n')
```

```python
user_table = []
properties_table = []
for i, unit in enumerate(['imperial', 'si']):

    user_table.append(
        user_table_prepare(
            pd.DataFrame(user_array[i]),
            unit
        )
    )

    properties_table.append(
        prop_table_units_handler(
            pd.DataFrame(properties_table_raw[i]),
            unit
        )
    )
```

```python
imperial = calculate_total_piping(Piping_Segment, Air, user_table[0], properties_table[0])
si = calculate_total_piping(Piping_Segment, Air, user_table[1], properties_table[1])

results_imperial = final_table_units_handler(imperial, 'imperial')
results_si = final_table_units_handler(si, 'imperial')
```
